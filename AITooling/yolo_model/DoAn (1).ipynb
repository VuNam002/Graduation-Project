```python
import os
import glob

# Đường dẫn cơ sở của bộ dữ liệu
base_dir = dataset.location
splits = ['train', 'valid', 'test']


# Tạo tiêu đề cho bảng thống kê
print(f"{'Tập (Split)':<12} | {'Số ảnh (Images)':<17} | {'Số file nhãn':<15} | {'Tổng số vật thể (BBox)'}")
print("-" * 70)

total_images = 0
total_labels = 0
total_bboxes = 0

# Quét qua từng tập dữ liệu
for split in splits:
    img_dir = os.path.join(base_dir, split, 'images')
    lbl_dir = os.path.join(base_dir, split, 'labels')

    num_imgs = 0
    num_lbls = 0
    num_bboxes = 0

    # Đếm số lượng ảnh
    if os.path.exists(img_dir):
        # Lấy các file phổ biến (jpg, jpeg, png)
        img_files = [f for f in os.listdir(img_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
        num_imgs = len(img_files)
        total_images += num_imgs

    # Đếm số lượng file nhãn và tổng số bounding box
    if os.path.exists(lbl_dir):
        lbl_files = glob.glob(os.path.join(lbl_dir, '*.txt'))
        num_lbls = len(lbl_files)
        total_labels += num_lbls

        for txt_file in lbl_files:
            with open(txt_file, 'r') as f:
                # Mỗi dòng trong file txt tương ứng với 1 bounding box
                num_bboxes += len(f.readlines())
        total_bboxes += num_bboxes

    # In dữ liệu của từng tập ra bảng
    print(f"{split.upper():<12} | {num_imgs:<17} | {num_lbls:<15} | {num_bboxes}")

print("-" * 70)
# In tổng cộng toàn bộ dataset
print(f"{'TỔNG CỘNG':<12} | {total_images:<17} | {total_labels:<15} | {total_bboxes}")
print("=" * 70)
```

```python
import os
import glob
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Thiết lập style chung cho các biểu đồ đẹp hơn
sns.set_theme(style="whitegrid")

# (Giả định bạn đã chạy code tải dataset từ Roboflow ở các cell trước đó)
base_dir = dataset.location
splits = ['train', 'valid', 'test']

print("Đang trích xuất dữ liệu từ các file YOLO txt...")
data_list = []

for split in splits:
    lbl_dir = os.path.join(base_dir, split, 'labels')
    if os.path.exists(lbl_dir):
        for txt_file in glob.glob(os.path.join(lbl_dir, '*.txt')):
            img_id = os.path.basename(txt_file).replace('.txt', '')
            with open(txt_file, 'r') as f:
                for line in f:
                    parts = line.strip().split()
                    if len(parts) == 5:
                        class_id, x_center, y_center, w, h = map(float, parts)
                        data_list.append([split, img_id, int(class_id), x_center, y_center, w, h])

# Tạo DataFrame
df = pd.DataFrame(data_list, columns=['Split', 'ImageID', 'ClassID', 'X_center', 'Y_center', 'Width', 'Height'])
df['ClassName'] = df['ClassID'].map(lambda x: class_names[x])
df['BBox_Area'] = df['Width'] * df['Height']
df['Aspect_Ratio'] = df['Width'] / df['Height']

print(f"Hoàn tất! Đã nạp thành công {len(df)} bounding boxes vào bộ nhớ.")
```

```python
import os
import cv2
import random
import math
import matplotlib.pyplot as plt

images_dir = os.path.join(dataset.location, "train/images")
labels_dir = os.path.join(dataset.location, "train/labels")

# Lấy tất cả các file hình ảnh hợp lệ
image_files = [f for f in os.listdir(images_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]

def show_random_images(n=12):
    # Tự động tính toán số hàng và số cột (ví dụ n=12 -> 3 hàng, 4 cột)
    cols = 4
    rows = math.ceil(n / cols)

    plt.figure(figsize=(18, 4 * rows))

    for i in range(n):
        img_file = random.choice(image_files)
        img_path = os.path.join(images_dir, img_file)

        # Cắt đuôi file cũ (.jpg, .png) và thay bằng .txt để tìm file nhãn
        base_name = os.path.splitext(img_file)[0]
        label_path = os.path.join(labels_dir, base_name + ".txt")

        img = cv2.imread(img_path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) # Chuyển ngay hệ màu để tránh quên
        h, w, _ = img.shape

        # Chỉ vẽ khung nếu file nhãn tồn tại
        if os.path.exists(label_path):
            with open(label_path) as f:
                lines = f.readlines()

            for line in lines:
                cls, x, y, bw, bh = map(float, line.split())

                # Giải mã tọa độ YOLO
                x1 = int((x - bw/2) * w)
                y1 = int((y - bh/2) * h)
                x2 = int((x + bw/2) * w)
                y2 = int((y + bh/2) * h)

                label = class_names[int(cls)]

                # Vẽ viền màu đỏ đậm (cho nổi bật) và dán tên class
                cv2.rectangle(img, (x1, y1), (x2, y2), (255, 0, 0), 2)
                cv2.putText(img, label, (x1, y1-8),
                            cv2.FONT_HERSHEY_SIMPLEX,
                            0.6, (255, 255, 0), 2)

        # Trực quan hóa lên lưới động
        plt.subplot(rows, cols, i+1)
        plt.imshow(img)
        plt.axis("off")
        plt.title(f"Ảnh {i+1}: {base_name[:10]}...", fontsize=10)

    plt.tight_layout()
    plt.show()

# Hiển thị lưới 12 ảnh (3 hàng x 4 cột)
show_random_images(12)
```

```python
import os
import glob
import yaml
import numpy as np
import pandas as pd

DATASET_PATH = "/content/Personal-Protective-Equipment---Combined-Model-8"

# Load class names
with open(os.path.join(DATASET_PATH, "data.yaml"), 'r') as f:
    data_yaml = yaml.safe_load(f)

class_names = data_yaml["names"]

# Load annotations
label_files = glob.glob(os.path.join(DATASET_PATH, "train/labels/*.txt"))

class_counts = {}

for file in label_files:
    with open(file, "r") as f:
        for line in f:
            class_id = int(line.split()[0])
            class_counts[class_id] = class_counts.get(class_id, 0) + 1

# Convert to dataframe
df_counts = pd.DataFrame({
    "class_id": list(class_counts.keys()),
    "count": list(class_counts.values())
})

df_counts["class_name"] = df_counts["class_id"].apply(lambda x: class_names[x])
df_counts = df_counts.sort_values("count", ascending=False)

df_counts
```

```python
max_count = df_counts["count"].max()
min_count = df_counts["count"].min()

imbalance_ratio = max_count / min_count

print("Imbalance Ratio:", imbalance_ratio)
```

```python
counts = df_counts["count"].values
probs = counts / counts.sum()

entropy = -np.sum(probs * np.log(probs))
max_entropy = np.log(len(probs))

normalized_entropy = entropy / max_entropy

print("Shannon Entropy:", entropy)
print("Max Possible Entropy:", max_entropy)
print("Normalized Entropy:", normalized_entropy)
```

```python
gini = 1 - np.sum(probs**2)
print("Gini Index:", gini)
```

```python
mean_count = np.mean(counts)
std_count = np.std(counts)

cv = std_count / mean_count

print("Coefficient of Variation:", cv)
```

```python
effective_classes = np.exp(entropy)

print("Effective Number of Classes:", effective_classes)
print("Actual Number of Classes:", len(counts))
```
