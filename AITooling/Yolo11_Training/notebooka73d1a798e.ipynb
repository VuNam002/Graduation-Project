```python
model = YOLO('yolo11n.pt')

results = model.train(
    data=yaml_path,
    epochs=100,
    patience=50,
    imgsz=640,
    batch=32,
    project='PPE_Project',
    name='Yolo11_Training',
    device=0,
    optimizer='auto',
    cos_lr=True
)
```

```python
import shutil
import os
from IPython.display import FileLink, display

# 1. Đường dẫn thư mục gốc
path_to_folder = '/kaggle/working/runs/detect/PPE_Project'
zip_name = 'PPE_Project_Export'

# 2. Tiến hành nén
if os.path.exists(path_to_folder):
    shutil.make_archive(zip_name, 'zip', path_to_folder)
    file_path = f"{zip_name}.zip"
    print(f"Nen thanh cong! Dang tao lien ket tai ve...")
    
    # 3. Tạo liên kết và tự động kích hoạt download bằng JavaScript
    display(FileLink(file_path))
    
    # Đoạn JS này sẽ tự động click vào liên kết vừa tạo
    from IPython.display import HTML
    shell_script = f"""
    <script>
        var link = document.createElement('a');
        link.href = '{file_path}';
        link.download = '{file_path}';
        document.body.appendChild(link);
        link.click();
        document.body.removeChild(link);
    </script>
    """
    display(HTML(shell_script))
    
else:
    print("Duong dan khong ton tai!")
```
